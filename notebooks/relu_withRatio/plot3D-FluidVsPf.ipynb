# 3D Surface: Output₁ vs Total Fluid & Propellant/Fluid Ratio

This notebook generates a static 3D surface showing how the model’s first output  
varying **Total Fluid** (X-axis) and **Propellant/Fluid ratio** (Y-axis), and  
highlights the actual sample point from your test set.


```python
import sys
from pathlib import Path

# Adjust PROJECT_ROOT to point at where your `src/` folder lives
PROJECT_ROOT = Path("/home/kamiar/chevron/Eagle-Ford")
sys.path.insert(0, str(PROJECT_ROOT / "src"))
```

```python
from my_package.pf_ratio_surfaces import plot_3d_response_pf_ratio_surface
```

```python
# Folder where cross_validate_mlp saved models & metadata
RUN_FOLDER               = "nnModelSaved_relu_withRatio_20250804"

# Which fold’s model to load
FOLD                     = 1

# Path to your test CSV (relative to this notebook)
TEST_CSV                 = "../../data/processed/Eagle Ford Data(Eagle Ford)_test.csv"

# Index in the test set to highlight (0-based)
SAMPLE_INDEX             = 1

# Sweep ranges
FLUID_RANGE              = (1e5, 2e7)    # min/max Total Fluid
PF_RATIO_RANGE           = (0.5, 1.5)    # min/max Propellant/Fluid ratio

# Grid resolution
N_FLUID                  = 30
N_RATIO                  = 30

# Whether to include the Proppant/gal‐per‐GPI ratio features
INCLUDE_RATIO_FEATURES   = True

# Optional: camera angle
ELEV                     = 30
AZIM                     = -60

# — New saving options —
SAVE_CSV                 = True   # write the grid data out to CSV
SAVE_PLOT                = True   # save the 3D figure as PNG
OUT_DIR                  = "plot3D-FluidVsPf_output"           # where to put CSV/PNG (will be created)
FILE_PREFIX              = "relu_withRatio_pf" # filename prefix for CSV/PNG
```

```python
plot_3d_response_pf_ratio_surface(
    run_folder             = RUN_FOLDER,
    fold                   = FOLD,
    test_csv               = TEST_CSV,
    sample_index           = SAMPLE_INDEX,
    fluid_range            = FLUID_RANGE,
    pf_ratio_range         = PF_RATIO_RANGE,
    n_fluid                = N_FLUID,
    n_ratio                = N_RATIO,
    include_ratio_features = INCLUDE_RATIO_FEATURES,
    elev                   = ELEV,
    azim                   = AZIM,
    save_csv               = SAVE_CSV,
    save_plot              = SAVE_PLOT,
    out_dir                = OUT_DIR,
    file_prefix            = FILE_PREFIX
)
```
