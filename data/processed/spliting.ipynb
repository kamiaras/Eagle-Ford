```python
import pandas as pd
from sklearn.model_selection import train_test_split
from pathlib import Path

# ─── Configuration ─────────────────────────────────────────────────────────────
# Path to raw CSV file (change as needed)
input_file = Path('../raw/Eagle Ford Data(Eagle Ford).csv')

# Fraction of data to hold out as the test set
test_frac = 0.15

# Seed for reproducibility of the split
random_state = 42
# ────────────────────────────────────────────────────────────────────────────────

# 1) Load the full dataset into a DataFrame
df = pd.read_csv(input_file)

# 2) Split into train+validation vs. test
train_val_df, test_df = train_test_split(
    df,
    test_size=test_frac,
    shuffle=True,
    random_state=random_state
)

# 3) Construct file paths for outputs
stem = input_file.stem  # e.g. "Eagle Ford Data(Eagle Ford)"
suffix = input_file.suffix  # e.g. ".csv"
train_val_path = input_file.parent / f"{stem}_train_val{suffix}"
test_path      = input_file.parent / f"{stem}_test{suffix}"

# 4) Save the splits (without adding an extra index column)
train_val_df.to_csv(train_val_path, index=False)
test_df.to_csv(test_path,      index=False)

# 5) Report file locations and row counts
print(f"Train+Validation ({len(train_val_df)} rows) → {train_val_path}")
print(f"Test             ({len(test_df)} rows) → {test_path}")
```
